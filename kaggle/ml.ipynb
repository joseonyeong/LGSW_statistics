## Scaling 변환
- (실제값 - 예측값)


```python
from sklearn.preprocessing import StandardScaler
import sklearn
import numpy as np

import pandas as pd

train = pd.read_csv('train.csv')
train.head()
```

```python
y = np.log1p(train['Calories'])
scaler = StandardScaler()



numeric_features = ['Age', 'Height', 'Weight', 'Duration', 'Heart_Rate', 'Body_Temp']
train[numeric_features].head(1)
```

```python
X = train[numeric_features]
X_scaled = scaler.fit_transform(X)
X_scaled[0]
```

```python
X.head(1)
```

```python
X_combined = pd.concat([
    pd.DataFrame(X_scaled, columns = numeric_features),
    train[['Sex_female','Sex_male']]
], axis = 1)

X_combined.head()
```

```python
X_combined.head()
```

# 학습/검증 데이터 분할
- 기본 원칙 : 층화추출 (비율에 따라 데이터셋 분리)
- 아래 코드는 층화추출이 된 상태가 아님 -> 무작위 샘플링
      vs 시계열                         -> 구간설정

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_trian, y_val = train_test_split(
    X_combined, y, test_size = 0.2, random_stat = 42        # random_stat 는 실험 재현성
)

X_trian.shape, X_val.shape, y_train.shape, y_val.shape      # y_train = target (종속변수)
```

# 모델링

```python
from sklearn.tree import DecisionTreeRegressor

dt_model = DecisionTreeRegressor(random_state=42)                   # 필수
dt_model.fit(X_train, y_train)                                      # 모델학습 끝
```
